```python
import numpy as np 
import xarray as xr
```

```python
ds = xr.open_dataset('swio_his.nc')
AKv = ds.AKv
AKs = ds.AKs
AKt = ds.AKt
s_rho = ds.s_rho
Cs_rho = ds.Cs_rho
hc = ds.hc
ds.close()
```

```python
fill_value = 9.96921e+36
AKv = AKv.where((AKv != fill_value), np.nan)
AKt = AKt.where((AKt != fill_value), np.nan)
AKs = AKs.where((AKs != fill_value), np.nan)
```

```python
ds = xr.Dataset({
    'AKt': AKt,
    'AKs': AKs,
    'AKv': AKv,
    's_rho': s_rho,
    'Cs_rho': Cs_rho,
    'hc': hc,
})
ds.to_netcdf('swio_his2.nc')
```
